```python
import networkx as nx
import numpy as np

# --- Fungsi untuk menampilkan representasi graf ---
def display_representations(graph, title):
    print(f"\n--- Representasi untuk {title} ---")

    # 1. Adjacency List
    adj_list = nx.to_dict_of_lists(graph)
    print("\nAdjacency List:")
    print(adj_list)

    # 2. Adjacency Matrix
    # Pastikan nodelist terurut untuk representasi matriks yang konsisten
    nodelist = sorted(list(graph.nodes()))
    adj_matrix = nx.to_numpy_array(graph, nodelist=nodelist)
    print("\nAdjacency Matrix:")
    print(adj_matrix)

    # 3. Incidence Matrix
    # NetworkX tidak memiliki fungsi bawaan untuk Incidence Matrix,
    # jadi kita akan membuatnya secara manual.
    nodes = list(graph.nodes())
    edges = list(graph.edges())
    num_nodes = len(nodes)
    num_edges = len(edges)

    incidence_matrix = np.zeros((num_nodes, num_edges), dtype=int)

    for i, node in enumerate(nodes):
        for j, edge in enumerate(edges):
            if node in edge:
                # Untuk graf tak berarah, setiap simpul yang merupakan endpoint sisi
                # akan memiliki nilai 1 di matriks ini.
                incidence_matrix[i, j] = 1

    print("\nIncidence Matrix (Rows = Nodes, Columns = Edges):")
    print(incidence_matrix)

# ==========================================================
# 1. Complete Graph (K_5)
# ==========================================================
K5 = nx.complete_graph(5)
display_representations(K5, "Complete Graph (K_5)")

# ==========================================================
# 2. Cycle Graph (C_6)
# ==========================================================
C6 = nx.cycle_graph(6)
display_representations(C6, "Cycle Graph (C_6)")

# ==========================================================
# 3. Wheel Graph (W_7)
# ==========================================================
W7 = nx.wheel_graph(7)
display_representations(W7, "Wheel Graph (W_7)")
```

# Task01_individual

Task01_Individu.py

1.	Buatlah sebuah graf Complete K4, graf Cycle C5, dan graf wheel W6. Visualisasikan ketiga graf tersebut.
2.	Untuk setiap graf, cetak properti berikut: jumlah simpul, jumlah sisi, dan derajat setiap simpul.


```python
import networkx as nx
import matplotlib.pyplot as plt

#fungsi untuk menampilkan properti grap
def display_graph_properties(graph, title):
    print(f"\n--- Properti untuk {title} ---")
    print(f"Jumlah Simpul (Nodes): {graph.number_of_nodes()}")
    print(f"Jumlah Sisi (Edges): {graph.number_of_edges()}")
    print("Derajat Simpul (Node Degrees):")
    for node, degree in graph.degree():
        print(f"  Simpul {node}: {degree}")

#Fungsi untuk visualisasi graf
def visualize_graph(graph, title):
    plt.figure(figsize=(6, 6))
    nx.draw(graph, with_labels=True, node_color='lightblue', edge_color='gray', node_size=700, font_size=10)
    plt.title(title)
    plt.show()

# ==========================================================
# 1. Complete Graph (K_4)
# ==========================================================
K4 = nx.complete_graph(4)
visualize_graph(K4, "Complete Graph (K_4)")
display_graph_properties(K4, "Complete Graph (K_4)")

# ==========================================================
# 2. Cycle Graph (C_5)
# ==========================================================
C5 = nx.cycle_graph(5)
visualize_graph(C5, "Cycle Graph (C_5)")
display_graph_properties(C5, "Cycle Graph (C_5)")

# ==========================================================
# 3. Wheel Graph (W_6)
# ==========================================================
W6 = nx.wheel_graph(6)
visualize_graph(W6, "Wheel Graph (W_6)")
display_graph_properties(W6, "Wheel Graph (W_6)")
```

# task02_individu

1.	Analisis graf: hitung derajat setiap simpul dan verifikasi Handshaking Theorem. (dari hasil task01)
2.	Representasikan graf ke dalam Adjacency List dan Adjacency Matrix.


```python
import networkx as nx
import numpy as np

# Function to verify Handshaking Theorem
def verify_handshaking_theorem(graph, title):
    total_degree = sum(dict(graph.degree()).values())
    num_edges = graph.number_of_edges()
    print(f"\n--- Verifikasi Handshaking Theorem untuk {title} ---")
    print(f"Jumlah total derajat: {total_degree}")
    print(f"2 * Jumlah sisi: {2 * num_edges}")
    if total_degree == 2 * num_edges:
        print("Handshaking Theorem terverifikasi.")
    else:
        print("Handshaking Theorem TIDAK terverifikasi.")

# Function to display graph representations
def display_representations(graph, title):
    print(f"\n--- Representasi untuk {title} ---")

    # 1. Adjacency List
    adj_list = nx.to_dict_of_lists(graph)
    print("\nAdjacency List:")
    print(adj_list)

    # 2. Adjacency Matrix
    nodelist = sorted(list(graph.nodes()))
    adj_matrix = nx.to_numpy_array(graph, nodelist=nodelist)
    print("\nAdjacency Matrix:")
    print(adj_matrix)

# ==========================================================
# Analyze and Represent Graphs (K4, C5, W6)
# ==========================================================

# 1. Complete Graph (K_4)
verify_handshaking_theorem(K4, "Complete Graph (K_4)")
display_representations(K4, "Complete Graph (K_4)")

# 2. Cycle Graph (C_5)
verify_handshaking_theorem(C5, "Cycle Graph (C_5)")
display_representations(C5, "Cycle Graph (C_5)")

# 3. Wheel Graph (W_6)
verify_handshaking_theorem(W6, "Wheel Graph (W_6)")
display_representations(W6, "Wheel Graph (W_6)")
```

# Task03_Timproyek

-	Skenario: Anda adalah manajer proyek dengan 4 developer (Alyn, Betty, Carol, David) dan 4 task (UI/UX, Backend, Database, Testing).
-	Keempat developer memiliki keahlian sebagai berikut: Alyn (UI/UX, Testing), Betty (UI/UX, Backend), Carol (UI/UX, Database), David (Backend, Testing).
-	Tugas:
	- Modelkan masalah ini sebagai Bipartite Graph.
  -	Representasikan graf ke dalam Adjacency List dan Adjacency Matrix.
	- Analisis graf: hitung derajat setiap simpul dan verifikasi Handshaking Theorem.
  - Identifikasi apakah ada isomorfisme antara graf Anda dengan graf khusus yang sudah dipelajari.


```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np

#data skenario
developers = ["Alyn", "Betty", "Carol", "David"]
tasks = ["UI/UX", "Backend", "Database", "Testing"]

# skill dev: (developer, task)
skills = [
    ("Alyn", "UI/UX"),
    ("Alyn", "Testing"),
    ("Betty", "UI/UX"),
    ("Betty", "Backend"),
    ("Carol", "UI/UX"),
    ("Carol", "Database"),
    ("David", "Backend"),
    ("David", "Testing")
]

# modelkn sbg bipartite graph
B = nx.Graph()
B.add_nodes_from(developers, bipartite=0) # set dev sbg 0
B.add_nodes_from(tasks, bipartite=1)      # set tasks sbg 1
B.add_edges_from(skills)

#visualisasi bipartite graph
plt.figure(figsize=(8, 6))
pos = nx.bipartite_layout(B, developers) # layout bipartite graph
nx.draw(B, pos, with_labels=True, node_color=['lightblue' if node in developers else 'lightgreen' for node in B.nodes()], edge_color='gray', node_size=700, font_size=10)
plt.title("Bipartite Graph: Developer Skills")
plt.show()

# representasi graf
print("\n--- Representasi Graf Bipartite ---")

# 1. Adjacency List
adj_list = nx.to_dict_of_lists(B)
print("\nAdjacency List:")
print(adj_list)

# 2. Adjacency Matrix
developers_sorted = sorted(developers)
tasks_sorted = sorted(tasks)

adj_matrix_bipartite = np.zeros((len(developers_sorted), len(tasks_sorted)), dtype=int)

for i, dev in enumerate(developers_sorted):
    for j, task in enumerate(tasks_sorted):
        if B.has_edge(dev, task):
            adj_matrix_bipartite[i, j] = 1

print("\nBipartite Adjacency Matrix (Rows = Developers, Columns = Tasks):")
print(adj_matrix_bipartite)

# analisis graf
print("\n--- Analisis Graf Bipartite ---")

# hitung derajat setiap simpul
print("\nDerajat Simpul (Node Degrees):")
for node, degree in B.degree():
    print(f"  Simpul {node}: {degree}")

# verifikasi handshaking theorem
total_degree = sum(dict(B.degree()).values())
num_edges = B.number_of_edges()
print(f"\nJumlah total derajat: {total_degree}")
print(f"2 * Jumlah sisi: {2 * num_edges}")
if total_degree == 2 * num_edges:
    print("Handshaking Theorem terverifikasi.")
else:
    print("Handshaking Theorem TIDAK terverifikasi.")

# Identifikasi Isomorfisme
print("\n--- Identifikasi Isomorfisme ---")
print("Untuk mengidentifikasi isomorfisme dengan graf khusus, kita perlu membandingkan properti graf ini (jumlah simpul, jumlah sisi, sebaran derajat, dll.) dengan properti graf khusus yang sudah dipelajari.")
print("Graf bipartite ini memiliki 8 simpul (4 developer + 4 task) dan 8 sisi.")
print("Sebaran derajat simpul developer: Alyn: 2, Betty: 2, Carol: 2, David: 2")
print("Sebaran derajat simpul task: UI/UX: 3, Backend: 2, Database: 1, Testing: 2")
print("\nBerdasarkan sebaran derajat yang tidak seragam dan struktur bipartite, graf ini mungkin tidak isomorfik dengan graf khusus seperti Complete Graph (K_n), Cycle Graph (C_n), atau Wheel Graph (W_n) yang biasanya memiliki pola derajat yang lebih seragam atau struktur yang berbeda.")
print("Identifikasi isomorfisme secara formal memerlukan algoritma khusus atau pemeriksaan properti yang lebih mendalam.")
```

# Task04_FullMesh

-	Skenario: Anda adalah administrator jaringan yang menganalisis topologi full-mesh dengan enam server yang terhubung langsung satu sama lain.
-	Tugas:
  - Modelkan ini sebagai Complete Graph (Kn) dan gambarkan.
  - Representasikan graf ini dalam Incidence Matrix.
  - Analisis graf: hitung derajat setiap simpul dan verifikasi Handshaking Theorem.
  - Jelaskan pengaruh kegagalan satu server pada connectivity graf.


```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np

#fungsi untuk menampilkan properti graph
def display_graph_properties(graph, title):
    print(f"\n--- Properti untuk {title} ---")
    print(f"Jumlah Simpul (Nodes): {graph.number_of_nodes()}")
    print(f"Jumlah Sisi (Edges): {graph.number_of_edges()}")
    print("Derajat Simpul (Node Degrees):")
    for node, degree in graph.degree():
        print(f"  Simpul {node}: {degree}")

# Fungsi untuk visualisasi graf
def visualize_graph(graph, title):
    plt.figure(figsize=(6, 6))
    nx.draw(graph, with_labels=True, node_color='lightblue', edge_color='gray', node_size=700, font_size=10)
    plt.title(title)
    plt.show()

# Fungsi untuk menampilkan Representasi Matriks Incidence
def display_incidence_matrix(graph, title):
    print(f"\n--- Representasi Matriks Incidence untuk {title} ---")
    nodes = list(graph.nodes())
    edges = list(graph.edges())
    num_nodes = len(nodes)
    num_edges = len(edges)

    incidence_matrix = np.zeros((num_nodes, num_edges), dtype=int)

    for i, node in enumerate(nodes):
        for j, edge in enumerate(edges):
            if node in edge:
                incidence_matrix[i, j] = 1

    print("\nIncidence Matrix (Rows = Nodes, Columns = Edges):")
    print(incidence_matrix)


# Modelkan sebagai Complete Graph (K_6)
num_servers = 6
K6 = nx.complete_graph(num_servers)

# Visualisasikan Graf
visualize_graph(K6, f"Full-Mesh Network ({num_servers} Servers - K_{num_servers})")

# Representasikan Graf dalam Incidence Matrix
display_incidence_matrix(K6, f"Full-Mesh Network (K_{num_servers})")

# Analisis Graf
display_graph_properties(K6, f"Full-Mesh Network (K_{num_servers})")

# Verifikasi Handshaking Theorem
total_degree = sum(dict(K6.degree()).values())
num_edges_k6 = K6.number_of_edges()
print(f"\n--- Verifikasi Handshaking Theorem untuk Full-Mesh Network (K_{num_servers}) ---")
print(f"Jumlah total derajat: {total_degree}")
print(f"2 * Jumlah sisi: {2 * num_edges_k6}")
if total_degree == 2 * num_edges_k6:
    print("Handshaking Theorem terverifikasi.")
else:
    print("Handshaking Theorem TIDAK terverifikasi.")

# Jelaskan pengaruh kegagalan satu server
print("\n--- Pengaruh Kegagalan Satu Server pada Konektivitas ---")
print("Dalam topologi full-mesh (Complete Graph), setiap server terhubung langsung ke setiap server lainnya.")
print("Jika satu server mengalami kegagalan:")
print("- Server yang gagal tidak dapat lagi berkomunikasi.")
print("- Semua server yang tersisa masih terhubung langsung satu sama lain.")
print("- Konektivitas antara server-server yang *tidak* gagal tetap utuh.")
print("Karena setiap pasangan server (yang tidak gagal) memiliki koneksi langsung, kegagalan satu server *tidak* akan memutus koneksi antara server-server yang tersisa. Graf yang tersisa (setelah menghapus simpul yang gagal dan sisi-sisinya) akan tetap menjadi Complete Graph dengan jumlah simpul satu lebih sedikit.")
print("Oleh karena itu, topologi full-mesh sangat tangguh terhadap kegagalan simpul tunggal dalam hal menjaga konektivitas antara simpul yang tersisa.")
```

# Task05_Datasister

-	Skenario: Anda menganalisis aliran data unidirectional antar 4 komponen: Processor, Memory, Storage, Network Interface.
-	Aliran data: Processor -> Memory, Processor -> Storage, Memory -> Processor, Storage -> Network Interface, Network Interface -> Processor.
-	Tugas:
  -	Modelkan ini sebagai Directed Graph.
  -	Representasikan graf ke Adjacency List dan Adjacency Matrix.
  -	Analisis graf: hitung in-degree dan out-degree setiap simpul dan verifikasi Handshaking Theorem versi berarah.
  -	Identifikasi isomorfisme dengan graf sederhana lain.


```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np

# data Skenario
components = ["Processor", "Memory", "Storage", "Network Interface"]

# data flow
data_flow = [
    ("Processor", "Memory"),
    ("Processor", "Storage"),
    ("Memory", "Processor"),
    ("Storage", "Network Interface"),
    ("Network Interface", "Processor")
]

# modelkan sebagai Directed Graph
D = nx.DiGraph()
D.add_nodes_from(components)
D.add_edges_from(data_flow)

# Visualisasi Graf Berarah
plt.figure(figsize=(8, 6))
pos = nx.spring_layout(D)
nx.draw(D, pos, with_labels=True, node_color='lightblue', edge_color='gray', node_size=1000, font_size=10, arrows=True, arrowstyle='->', arrowsize=20)
plt.title("Directed Graph: Data Flow")
plt.show()

print("\n--- Representasi Graf Berarah ---")

# 1. Adjacency List
adj_list_directed = nx.to_dict_of_lists(D)
print("\nAdjacency List (Directed):")
print(adj_list_directed)

# 2. Adjacency Matrix
nodelist_sorted = sorted(list(D.nodes()))
adj_matrix_directed = nx.to_numpy_array(D, nodelist=nodelist_sorted)
print("\nAdjacency Matrix (Directed):")
print(adj_matrix_directed)

# --- Analisis Graf Berarah ---
print("\n--- Analisis Graf Berarah ---")

# Hitung in-degree dan out-degree setiap simpul
print("\nDerajat Simpul (Node Degrees - Directed):")
for node in D.nodes():
    in_degree = D.in_degree(node)
    out_degree = D.out_degree(node)
    print(f"  Simpul {node}: In-Degree = {in_degree}, Out-Degree = {out_degree}")

# Verifikasi Handshaking Theorem versi Berarah
sum_in_degrees = sum(dict(D.in_degree()).values())
sum_out_degrees = sum(dict(D.out_degree()).values())
num_edges_directed = D.number_of_edges()

print(f"\n--- Verifikasi Handshaking Theorem untuk Graf Berarah ---")
print(f"Jumlah total In-Degree: {sum_in_degrees}")
print(f"Jumlah total Out-Degree: {sum_out_degrees}")
print(f"Jumlah sisi: {num_edges_directed}")

if sum_in_degrees == sum_out_degrees and sum_in_degrees == num_edges_directed:
    print("Handshaking Theorem terverifikasi.")
else:
    print("Handshaking Theorem TIDAK terverifikasi.")

# --- Identifikasi Isomorfisme ---
print("\n--- Identifikasi Isomorfisme dengan Graf Sederhana Lain ---")
print("Untuk mengidentifikasi isomorfisme dengan graf sederhana lain, kita perlu membandingkan properti graf berarah ini, seperti jumlah simpul, jumlah sisi, sebaran in-degree dan out-degree, dan keberadaan siklus, dengan properti graf sederhana berarah lainnya.")
print("Graf ini memiliki 4 simpul dan 5 sisi.")
print("Sebaran In-Degree: Processor: 2, Memory: 1, Storage: 1, Network Interface: 1")
print("Sebaran Out-Degree: Processor: 2, Memory: 1, Storage: 1, Network Interface: 1")
print("\nBerdasarkan sebaran derajat yang tidak seragam dan strukturnya yang spesifik, graf ini mungkin tidak isomorfik dengan graf berarah sederhana yang umum (misalnya, siklus berarah atau graf lengkap berarah) yang memiliki pola derajat yang lebih teratur.")
print("Identifikasi isomorfisme secara formal memerlukan algoritma khusus untuk graf berarah.")
```
