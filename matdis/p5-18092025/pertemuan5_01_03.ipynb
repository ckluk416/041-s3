# Task01
1.	Mahasiswa diminta untuk menjalankan dan menganalisis kode dibawah ini
2.	Tugas individu ini bertujuan untuk memastikan setiap mahasiswa memahami dasar implementasi euler path, hamilton dan greedy algorithm.


```python
import networkx as nx
import matplotlib.pyplot as plt
import random

# --- Bagian 1: Euler Path dan Circuit ---
print("--- Bagian 1: Euler Path dan Circuit ---")
# Contoh graf dengan Euler Circuit (semua simpul berderajat genap)
G_euler_circuit = nx.Graph([('A', 'B'), ('A', 'C'), ('B', 'C'), ('B', 'D'), ('C', 'D')])
print("Derajat simpul pada graf G_euler_circuit:")
for node in G_euler_circuit.nodes():
    print(f"Simpul {node}: {G_euler_circuit.degree(node)}")
print(f"Apakah memiliki Euler Circuit? {nx.is_eulerian(G_euler_circuit)}")
# Euler Circuit tidak tersedia di NetworkX, perlu implementasi manual atau library lain
print("Visualisasi G_euler_circuit:")
pos = nx.spring_layout(G_euler_circuit, seed=42)
nx.draw(G_euler_circuit, pos, with_labels=True, node_color='lightgreen', font_weight='bold')
plt.title("Graf dengan Euler Circuit")
plt.show()

print("\n--- Bagian 2: Hamilton Path dan Circuit ---")
# Contoh graf dengan Hamilton Circuit (Cn)
G_hamilton_circuit = nx.cycle_graph(5)
print("Visualisasi G_hamilton_circuit:")
pos = nx.circular_layout(G_hamilton_circuit)
nx.draw(G_hamilton_circuit, pos, with_labels=True, node_color='lightblue', font_weight='bold')
plt.title("Graf dengan Hamilton Circuit (C5)")
plt.show()
# Hamilton Path/Circuit juga tidak tersedia di NetworkX, karena ini masalah yang sulit.
# Kita hanya akan memvisualisasikan contohnya.

print("\n--- Bagian 3: Algoritma Dijkstra ---")
# Contoh graf berbobot untuk Dijkstra
G_dijkstra = nx.Graph()
edges = [
    ('A', 'B', 4), ('A', 'C', 2), ('B', 'C', 1),
    ('B', 'D', 5), ('C', 'D', 8), ('C', 'E', 10),
    ('D', 'E', 2), ('D', 'F', 6), ('E', 'F', 3)
]
G_dijkstra.add_weighted_edges_from(edges)
print("Graf berbobot untuk Dijkstra:")
pos = nx.planar_layout(G_dijkstra)
nx.draw(G_dijkstra, pos, with_labels=True, node_color='lightyellow', font_weight='bold', node_size=700)
edge_labels = nx.get_edge_attributes(G_dijkstra, 'weight')
nx.draw_networkx_edge_labels(G_dijkstra, pos, edge_labels=edge_labels)
plt.title("Graf Berbobot")
plt.show()

# Mencari jalur terpendek dari 'A' ke 'F'
shortest_path = nx.dijkstra_path(G_dijkstra, source='A', target='F')
shortest_path_length = nx.dijkstra_path_length(G_dijkstra, source='A', target='F')

print(f"Jalur terpendek dari A ke F: {shortest_path}")
print(f"Panjang jalur terpendek: {shortest_path_length}")

# Visualisasi jalur terpendek
path_edges = list(zip(shortest_path, shortest_path[1:]))
nx.draw(G_dijkstra, pos, with_labels=True, node_color='lightyellow', font_weight='bold', node_size=700)
nx.draw_networkx_edges(G_dijkstra, pos, edgelist=path_edges, edge_color='red', width=2)
nx.draw_networkx_edge_labels(G_dijkstra, pos, edge_labels=edge_labels)
plt.title("Jalur Terpendek dari A ke F")
plt.show()
```

# task02
Task02_ Rute_postman.py
- Skenario: Anda adalah seorang kepala logistik. Anda perlu merancang rute untuk kurir pengantar paket agar dapat melewati setiap jalan di sebuah area perumahan tepat satu kali.
- Tugas:
  - Modelkan area perumahan ini sebagai sebuah graf tak berarah. Graf harus memiliki 8 simpul dan 12 sisi, di mana semua simpul berderajat genap.
  - Verifikasi bahwa graf tersebut memenuhi syarat untuk Euler Circuit menggunakan NetworkX.
  - Jelaskan mengapa Euler Circuit lebih cocok untuk masalah ini dibandingkan Hamilton Circuit.
  - Visualisasikan graf dan tunjukkan derajat dari setiap simpul.


```python
import networkx as nx
import matplotlib.pyplot as plt

print("--- Task 02: Rute Postman ---")
G_postman = nx.Graph()

nodes = list(range(1, 9))
G_postman.add_nodes_from(nodes)

edges_postman = [
    (1, 3), (1, 4), (1, 5), (1, 6), (1, 7), (1, 8),
    (2, 3), (2, 4), (2, 5), (2, 6), (2, 7), (2, 8)
]


G_postman.add_edges_from(edges_postman)

print("Graf untuk Task 02:")
print(f"Jumlah simpul: {G_postman.number_of_nodes()}")
print(f"Jumlah sisi: {G_postman.number_of_edges()}")

print("\nDerajat simpul pada graf G_postman:")
for node in G_postman.nodes():
    print(f"Simpul {node}: {G_postman.degree(node)}")

is_connected = nx.is_connected(G_postman)
is_eulerian = nx.is_eulerian(G_postman)

print(f"\nApakah graf terhubung? {is_connected}")
print(f"Apakah memiliki Euler Circuit? {is_eulerian}")

print("\nVisualisasi Graf untuk Task 02:")
pos_postman = nx.spring_layout(G_postman, seed=42)
nx.draw(G_postman, pos_postman, with_labels=True, node_color='salmon', font_weight='bold', node_size=700)

node_degree_labels = {node: G_postman.degree(node) for node in G_postman.nodes()}
pos_labels = {}
for node, coords in pos_postman.items():
    pos_labels[node] = (coords[0], coords[1] + 0.1)

nx.draw_networkx_labels(G_postman, pos_labels, labels=node_degree_labels, font_size=10, verticalalignment='bottom')

plt.title("Graf Area Perumahan (Task 02) dengan Derajat Simpul")
plt.show()
```

# task03
Task03_ Rencana Perjalanan Salesperson.py
- Skenario: Seorang salesperson perlu mengunjungi beberapa kota untuk bertemu klien. Tugas Anda adalah menemukan jalur tercepat dari kota asal ke kota tujuan.
-	Untuk membuat graf, cari dari wikipedia tokoh masyarakat yang merupakan lulusan dari universitas tersebut baik S1, S2 maupun S3, masing-masing universitas 3 orang (total 12 vertex)
-	Tugas:
  - Buatlah sebuah graf berbobot tak berarah yang merepresentasikan 6 kota dan jalur penerbangannya. Bobot pada sisi adalah waktu tempuh dalam jam.
  - Kota: Bandung, Jakarta, Surabaya, Yogyakarta, Bali, Lombok.
  - Jalur & Waktu:
    - Bandung-Jakarta (1 jam)
    - Jakarta-Surabaya (2 jam)
    - Jakarta-Yogyakarta (1.5 jam)
    - Surabaya-Bali (2 jam)
    - Yogyakarta-Bali (3 jam)
    - Bali-Lombok (0.5 jam)
    - Jakarta-Lombok (3.5 jam)
  - Gunakan algoritma Dijkstra untuk menemukan jalur penerbangan tercepat dari Bandung ke Lombok.
  - Tampilkan jalur dan total waktu tempuh tercepat.
  - Visualisasikan graf beserta jalur terpendeknya dengan warna yang berbeda.


```python
import networkx as nx
import matplotlib.pyplot as plt

print("--- Task 03: Rencana Perjalanan Salesperson ---")

G_salesperson = nx.Graph()

cities = ["Bandung", "Jakarta", "Surabaya", "Yogyakarta", "Bali", "Lombok"]
G_salesperson.add_nodes_from(cities)

edges_salesperson = [
    ("Bandung", "Jakarta", 1),
    ("Jakarta", "Surabaya", 2),
    ("Jakarta", "Yogyakarta", 1.5),
    ("Surabaya", "Bali", 2),
    ("Yogyakarta", "Bali", 3),
    ("Bali", "Lombok", 0.5),
    ("Jakarta", "Lombok", 3.5)
]
G_salesperson.add_weighted_edges_from(edges_salesperson)

print("Graf Jalur Penerbangan:")
print(f"Jumlah simpul (kota): {G_salesperson.number_of_nodes()}")
print(f"Jumlah sisi (jalur): {G_salesperson.number_of_edges()}")


shortest_path_salesperson = nx.dijkstra_path(G_salesperson, source='Bandung', target='Lombok')
shortest_time_salesperson = nx.dijkstra_path_length(G_salesperson, source='Bandung', target='Lombok')

print(f"\nJalur penerbangan tercepat dari Bandung ke Lombok: {shortest_path_salesperson}")
print(f"Total waktu tempuh tercepat: {shortest_time_salesperson} jam")

print("\nVisualisasi Graf Jalur Penerbangan dengan Jalur Terpendek:")
pos_salesperson = nx.spring_layout(G_salesperson, seed=42)
nx.draw(G_salesperson, pos_salesperson, with_labels=True, node_color='lightcoral', font_weight='bold', node_size=700)

edge_labels_salesperson = nx.get_edge_attributes(G_salesperson, 'weight')
nx.draw_networkx_edge_labels(G_salesperson, pos_salesperson, edge_labels=edge_labels_salesperson)

path_edges_salesperson = list(zip(shortest_path_salesperson, shortest_path_salesperson[1:]))
nx.draw_networkx_edges(G_salesperson, pos_salesperson, edgelist=path_edges_salesperson, edge_color='blue', width=2)

plt.title("Graf Jalur Penerbangan dengan Jalur Terpendek")
plt.show()
```
