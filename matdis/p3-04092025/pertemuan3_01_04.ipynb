# Tutorial01.py/Task01_Individu.py
Mahasiswa dapat menjalankan skrip Python di bawah ini untuk memahami konsep-konsep dasar connectivity secara visual. Kode ini menunjukkan bagaimana mendefinisikan path dan circuit, serta bagaimana NetworkX menganalisis keterhubungan, komponen terhubung, cut vertex, dan cut edge.

task01

  1.	Mahasiswa diminta untuk menjalankan dan menganalisis kode pada modul_praktikum_minggu_3 tutorial.py
  2.	Tugas individu ini bertujuan untuk memastikan setiap mahasiswa memahami dasar-dasar konektivitas sebelum masuk ke studi kasus kolaboratif


```python
import networkx as nx
import matplotlib.pyplot as plt

# Fungsi untuk memvisualisasikan graf
def visualize_graph(G, title, pos=None, highlight_edges=None, highlight_nodes=None):
    plt.figure(figsize=(8, 6))
    if pos is None:
        if isinstance(G, nx.DiGraph):
            pos = nx.spring_layout(G, seed=42)
        else:
            pos = nx.spring_layout(G, seed=42)

    # Menggambar semua simpul
    nx.draw_networkx_nodes(G, pos, node_color='skyblue', node_size=700)

    # Menggambar semua sisi
    nx.draw_networkx_edges(G, pos, edge_color='gray')

    # Menggambar simpul yang di-highlight
    if highlight_nodes:
        nx.draw_networkx_nodes(G, pos, nodelist=highlight_nodes, node_color='red', node_size=700)

    # Menggambar sisi yang di-highlight
    if highlight_edges:
        nx.draw_networkx_edges(G, pos, edgelist=highlight_edges, edge_color='red', width=2)

    # Menggambar label
    nx.draw_networkx_labels(G, pos, font_size=12, font_family="sans-serif")

    plt.title(title)
    plt.axis('off')
    plt.show()

# --- Materi: Path dan Circuit ---
print("--- Materi: Path dan Circuit ---")
# Contoh graf sederhana
G_path = nx.Graph()
G_path.add_edges_from([('a', 'b'), ('b', 'c'), ('c', 'd'), ('d', 'e'), ('e', 'a'), ('b', 'd')])
visualize_graph(G_path, "Contoh Graf dengan Path dan Circuit")

# Path dari a ke e: a-b-c-d-e
path_nodes = ['a', 'b', 'c', 'd', 'e']
path_edges = [('a', 'b'), ('b', 'c'), ('c', 'd'), ('d', 'e')]
print("Path dari a ke e: a -> b -> c -> d -> e")
visualize_graph(G_path, "Visualisasi Path", highlight_edges=path_edges)

# Circuit: a-b-d-c-a
circuit_nodes = ['a', 'b', 'd', 'c', 'a']
circuit_edges = [('a', 'b'), ('b', 'd'), ('d', 'c'), ('c', 'a')]
print("\nCircuit: a -> b -> d -> c -> a")
visualize_graph(G_path, "Visualisasi Circuit", highlight_edges=circuit_edges)


print("\n--- Materi: Keterhubungan Graf Tak Berarah ---")
# Contoh graf terhubung dan tidak terhubung
G_connected = nx.Graph()
G_connected.add_edges_from([('1', '2'), ('2', '3'), ('3', '4'), ('4', '1'), ('1', '3')])
visualize_graph(G_connected, "Graf Terhubung")
print(f"Apakah graf terhubung? {nx.is_connected(G_connected)}")

G_disconnected = nx.Graph()
G_disconnected.add_edges_from([('1', '2'), ('3', '4'), ('5', '6')])
visualize_graph(G_disconnected, "Graf Tidak Terhubung")
print(f"Apakah graf terhubung? {nx.is_connected(G_disconnected)}")
print(f"Jumlah komponen terhubung: {nx.number_connected_components(G_disconnected)}")

# Contoh Cut Vertex dan Cut Edge
G_cut = nx.Graph()
G_cut.add_edges_from([('a', 'b'), ('b', 'c'), ('c', 'd'), ('d', 'e'), ('e', 'f'), ('f', 'g'), ('g', 'd')])
visualize_graph(G_cut, "Contoh Cut Vertex dan Cut Edge")

# Menemukan cut vertices
cut_vertices = list(nx.articulation_points(G_cut))
print(f"Cut Vertices (Titik Potong): {cut_vertices}")
# Visualisasi dengan menghapus cut vertex 'd'
G_cut_d = G_cut.copy()
G_cut_d.remove_node('d')
visualize_graph(G_cut_d, "Graf setelah menghapus Cut Vertex 'd'")

# Menemukan cut edges
cut_edges = list(nx.bridges(G_cut))
print(f"Cut Edges (Sisi Potong): {cut_edges}")
# Visualisasi dengan menghapus cut edge ('a', 'b')
G_cut_ab = G_cut.copy()
G_cut_ab.remove_edge('a', 'b')
visualize_graph(G_cut_ab, "Graf setelah menghapus Cut Edge ('a', 'b')")

print("\n--- Latihan Praktikum Minggu 3 ---")

# Latihan 1: Analisis Keterhubungan Graf Tak Berarah
G_latihan_1 = nx.Graph()
G_latihan_1.add_edges_from([('1', '2'), ('1', '3'), ('2', '3'), ('3', '4'), ('4', '5'), ('5', '6'), ('4', '6')])
visualize_graph(G_latihan_1, "Latihan 1: Analisis Keterhubungan")
print("\nLatihan 1:")
print(f"Apakah graf terhubung? {nx.is_connected(G_latihan_1)}")
print(f"Cut Vertices (Titik Potong): {list(nx.articulation_points(G_latihan_1))}")
print(f"Cut Edges (Sisi Potong): {list(nx.bridges(G_latihan_1))}")

# Latihan 2: Keterhubungan Graf Berarah
G_latihan_2 = nx.DiGraph()
G_latihan_2.add_edges_from([('a', 'b'), ('b', 'c'), ('c', 'a'), ('a', 'd')])
visualize_graph(G_latihan_2, "Latihan 2: Keterhubungan Graf Berarah")
print("\nLatihan 2:")
print(f"Apakah graf terhubung kuat? {nx.is_strongly_connected(G_latihan_2)}")
print(f"Apakah graf terhubung lemah? {nx.is_weakly_connected(G_latihan_2)}")
# Menemukan komponen yang terhubung kuat
strong_components = list(nx.strongly_connected_components(G_latihan_2))
print(f"Strongly Connected Components: {strong_components}")
```

# Task02_ Analisis Jaringan Jalan Raya dan Kemacetan.py
-	Skenario: Anda adalah seorang analis transportasi di Bandung. Anda ingin memodelkan kemacetan di persimpangan jalan Soekarno-Hatta (Simpang Lima). merupakan persimpangan terbanyak paling populer di Kota Bandung. Simpang ini adalah pertemuan lima jalan, yaitu Jl. Asia Afrika, Jl. Sunda, Jl. Sunda–Karapitan, Jl. Gatot Subroto, dan Jl. Ahmad Yani. . Untuk mempermudah, asumsikan persimpangan ini adalah sebuah graf, di mana simpul adalah persimpangan dan sisi adalah jalan.
-	Tugas:
  - Search simpang lima di google maps (print screen simpang lima)
  -	Modelkan persimpangan ini sebagai graf tak berarah.
  - Identifikasi Cut Vertex dan Cut Edge pada graf tersebut.
  - Jelaskan makna Cut Vertex dan Cut Edge dalam konteks jaringan jalan raya di persimpangan tersebut. Mengapa persimpangan tersebut sering mengalami kemacetan? Kaitkan dengan konsep Cut Vertex.
  - Visualisasikan graf, lalu visualisasikan kembali setelah salah satu Cut Vertex dihapus untuk menunjukkan bahwa graf menjadi tidak terhubung.



```python
import networkx as nx
import matplotlib.pyplot as plt

# Fungsi untuk memvisualisasikan graf
def visualize_graph(G, title, pos=None, highlight_edges=None, highlight_nodes=None):
    plt.figure(figsize=(8, 6))

    if pos is None:
        pos = nx.circular_layout(G)

    nx.draw_networkx_nodes(G, pos, node_color='skyblue', node_size=700)
    nx.draw_networkx_edges(G, pos, edge_color='gray')
    if highlight_nodes:
        nx.draw_networkx_nodes(G, pos, nodelist=highlight_nodes, node_color='red', node_size=700)
    if highlight_edges:
        nx.draw_networkx_edges(G, pos, edgelist=highlight_edges, edge_color='red', width=2)
    nx.draw_networkx_labels(G, pos, font_size=12, font_family="sans-serif")

    plt.title(title)
    plt.axis('off')
    plt.show()


# Definisi graf sesuai file
G_path = nx.Graph()
G_path.add_edges_from([
    ('Jl.Asia Afrika', 'Jl. Sunda'),
    ('Jl. Karapitan', 'Jl. A.Yani'),
    ('Jl. Karapitan', 'Jl. Sunda'),
    ('Jl. Karapitan', 'Jl. Gatot Subroto'),
    ('Jl.Asia Afrika', 'Jl. A.Yani'),
    ('Jl. A.Yani', 'Jl. Gatot Subroto'),
    ('Jl. Gatot Subroto',  'Jl. Sunda')
])

pos_custom = {
    'Jl. Karapitan': (0, 2),
    'Jl. A.Yani': (2, 1.5),
    'Jl. Gatot Subroto': (2, -1),
    'Jl. Sunda': (-1, -1),
    'Jl.Asia Afrika': (-2, 0.5),
}

# Visualisasi graf asli
visualize_graph(G_path, "Graf Simpang Lima", pos=pos_custom)

# Visualisasi setelah hapus cut vertices
G_cut_d = G_path.copy()
removed_nodes = ['Jl. Sunda', 'Jl. A.Yani']
G_cut_d.remove_nodes_from(removed_nodes)
visualize_graph(G_cut_d, "Graf setelah Cut Vertices 'Jl. Sunda' dan 'Jl. A.Yani'", pos=pos_custom)

# Visualisasi setelah hapus cut edges
G_cut_ab = G_path.copy()
removed_edges = [
    ('Jl.Asia Afrika', 'Jl. Sunda'),
    ('Jl.Asia Afrika', 'Jl. A.Yani')
]
G_cut_ab.remove_edges_from(removed_edges)
visualize_graph(G_cut_ab, "Graf setelah Cut Edges", pos=pos_custom)

# === Hasil Analisis ===
print("\n=== Hasil Analisis Graf Simpang Lima ===")
print(f"Simpul yang dihapus (Cut Vertices): {removed_nodes}")
print(f"Sisa simpul setelah penghapusan: {list(G_cut_d.nodes)}")
print(f"Sisi yang dihapus (Cut Edges): {removed_edges}")
print(f"Sisa sisi setelah penghapusan: {list(G_cut_ab.edges())}")

print("\n--- Interpretasi Lalu Lintas ---")
print("""
1. Cut Vertex (Titik Kritis Simpul):
   - 'Jl. Sunda':
     Jika simpang ini terganggu (misalnya karena perbaikan jalan, lampu lalu lintas mati, atau kecelakaan),
     maka konektivitas Asia Afrika ke jalur lain berkurang drastis. Akibatnya, arus kendaraan dari Asia Afrika
     harus berputar jauh atau menumpuk di jalur A.Yani. Hal ini menimbulkan kemacetan parah di sekitar simpang.

   - 'Jl. A.Yani':
     Jika simpang ini tidak berfungsi, Asia Afrika tidak bisa lagi mengakses Gatot Subroto secara langsung.
     Seluruh arus yang biasanya melewati A.Yani harus dialihkan ke Jl. Sunda, yang tidak didesain menampung
     volume sebesar itu. Akibatnya terjadi bottleneck dan kemacetan panjang di ruas Sunda dan sekitarnya.

2. Cut Edge (Sisi Kritis Ruas Jalan):
   - ('Jl.Asia Afrika', 'Jl. Sunda'):
     Ruas ini adalah salah satu jalur utama keluar-masuk Asia Afrika. Jika ditutup, Asia Afrika hanya bisa
     diakses lewat A.Yani. Seluruh beban lalu lintas dialihkan ke satu jalur saja → kepadatan meningkat signifikan.

   - ('Jl.Asia Afrika', 'Jl. A.Yani'):
     Sama halnya, ruas ini adalah jalur utama lain. Jika ruas ini terputus, Asia Afrika hanya bisa keluar lewat Sunda.
     Ini membuat jalur Sunda menjadi satu-satunya akses, sehingga arus kendaraan menumpuk di simpang Sunda.

3. Kesimpulan:
   - Jl. Sunda dan Jl. A.Yani adalah simpul paling kritis (cut vertices).
   - Ruas AsiaAfrika–Sunda dan AsiaAfrika–A.Yani adalah sisi paling kritis (cut edges).
   - Dalam konteks nyata, titik-titik inilah yang rawan menyebabkan kemacetan besar di Simpang Lima, karena
     jika satu saja terganggu, jaringan jalan tidak lagi terhubung dengan baik, dan arus kendaraan menumpuk di titik tersisa.
""")


```

# Task03_ Analisis Jejaring Sosial.py
-	Skenario: Anda mengamati jejaring sosial antara para tokoh publik, di mana pertemanan diwakili oleh simpul, dan afiliasi dengan universitas (UGM, ITB, UI, Unpad) membentuk kelompok-kelompok. Tokoh-tokoh ini tidak selalu saling mengenal, sehingga jejaring ini memiliki beberapa connected components.
-	Untuk membuat graf, cari dari wikipedia tokoh masyarakat yang merupakan lulusan dari universitas tersebut baik S1, S2 maupun S3, masing-masing universitas 3 orang (total 12 vertex)
-	Tugas:
  - Buatlah sebuah graf tak berarah yang memiliki setidaknya tiga connected components. Setiap komponen harus merepresentasikan sebuah afiliasi universitas dan memiliki minimal 3 simpul (tokoh).
  - Tampilkan jumlah connected components dan daftar simpul di setiap komponen.
  - Jelaskan makna connected components dalam konteks jejaring sosial ini. Kaitkan dengan asumsi bahwa pertemanan cenderung terjadi di dalam lingkungan afiliasi yang sama.
  - Visualisasikan graf Anda, lalu gunakan warna yang berbeda untuk setiap connected component untuk memperjelas hasilnya.


```python
import networkx as nx
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

# UGM (Universitas Gadjah Mada)
ugm_alumni = [
    "Pratikno",
    "Adrianus Mooy",
    "J Soedrajad Djiwandono"
]

# ITB (Institut Teknologi Bandung)
itb_alumni = [
    "Ignasius Jonan",
    "Ridwan Kamil",
    "Hetifah Sjaifudian"
]

# UI (Universitas Indonesia)
ui_alumni = [
    "Boediono",
    "Jusuf Wanandi",
    "Mohammad Hatta"
]

# Unpad (Universitas Padjadjaran)
unpad_alumni = [
    "Ahmad Heryawan",
    "Hikmat Kurnia",
    "Ernest Prakasa"
]

# Gabungkan semua alumni menjadi list nodes
all_alumni = ugm_alumni + itb_alumni + ui_alumni + unpad_alumni

# Buat graf tak berarah
G = nx.Graph()

# Tambahkan semua nodes (tokoh)
G.add_nodes_from(all_alumni)

# Tambahkan edges hanya di dalam kelompok universitas yang sama
# Untuk UGM
for i in range(len(ugm_alumni)):
    for j in range(i + 1, len(ugm_alumni)):
        G.add_edge(ugm_alumni[i], ugm_alumni[j])

# Untuk ITB
for i in range(len(itb_alumni)):
    for j in range(i + 1, len(itb_alumni)):
        G.add_edge(itb_alumni[i], itb_alumni[j])

# Untuk UI
for i in range(len(ui_alumni)):
    for j in range(i + 1, len(ui_alumni)):
        G.add_edge(ui_alumni[i], ui_alumni[j])

# Untuk Unpad
for i in range(len(unpad_alumni)):
    for j in range(i + 1, len(unpad_alumni)):
        G.add_edge(unpad_alumni[i], unpad_alumni[j])

# Tambahkan edges antar universitas berdasarkan relasi nyata di dunia nyata
# UGM - ITB
G.add_edge("Pratikno", "Ignasius Jonan")
G.add_edge("Pratikno", "Ridwan Kamil")
G.add_edge("Pratikno", "Hetifah Sjaifudian")

# ITB - Unpad
G.add_edge("Ridwan Kamil", "Ahmad Heryawan")
G.add_edge("Ridwan Kamil", "Ernest Prakasa")
G.add_edge("Ridwan Kamil", "Hikmat Kurnia")

# Unpad - UI
G.add_edge("Ahmad Heryawan", "Boediono")
G.add_edge("Ahmad Heryawan", "Mohammad Hatta")
G.add_edge("Ahmad Heryawan", "Jusuf Wanandi")

# UI - UGM
G.add_edge("Boediono", "Pratikno")
G.add_edge("Boediono", "Adrianus Mooy")
G.add_edge("Boediono", "J Soedrajad Djiwandono")

# Hitung connected components setelah update
components = list(nx.connected_components(G))
num_components = len(components)

# Tampilkan jumlah connected components dan daftar simpul di setiap komponen
print("Jumlah connected components setelah analisis relasi nyata:", num_components)
for idx, comp in enumerate(components, 1):
    print(f"Connected Component {idx}: {list(comp)}")

# Penjelasan awal
print("\nPenjelasan Awal:")
print("Connected components dalam konteks jejaring sosial ini merepresentasikan kelompok-kelompok tokoh publik yang terhubung melalui pertemanan atau interaksi sosial yang diasumsikan terjadi di dalam lingkungan afiliasi universitas yang sama. Setiap komponen adalah sekelompok individu dari universitas yang sama yang saling terhubung (misalnya, melalui pertemanan), tetapi awalnya tidak terhubung dengan kelompok dari universitas lain. Hal ini menggambarkan asumsi homophily, di mana pertemanan cenderung terbentuk di antara individu dengan latar belakang atau afiliasi yang serupa, seperti alumni dari universitas yang sama, sehingga membentuk komunitas terpisah dalam jejaring sosial yang lebih besar.")

# Analisis relasi antar universitas
print("\nAnalisis Keterhubungan Nyata Antar Universitas:")
print("Berdasarkan pencarian dari sumber terpercaya, keterhubungan nyata antar tokoh dari universitas berbeda terutama terjadi melalui kolaborasi politik, pemerintahan, dan kebijakan ekonomi. Contoh:")
print("- Pratikno (UGM) dan Ignasius Jonan (ITB): Kolega di kabinet Presiden Jokowi, di mana Pratikno menjabat Menteri Sekretariat Negara dan Jonan sebagai Menteri ESDM/Transportasi, terlibat dalam pertemuan kabinet untuk isu energi dan transportasi.")
print("- Ridwan Kamil (ITB) dan Ahmad Heryawan (Unpad): Ridwan Kamil adalah penerus langsung Ahmad Heryawan sebagai Gubernur Jawa Barat (2018-2023 setelah 2008-2018), menunjukkan suksesi politik di provinsi yang sama.")
print("- Boediono (UI) dan Adrianus Mooy (UGM): Keduanya terlibat dalam teknokrasi ekonomi Indonesia; Mooy sebagai mantan Gubernur BI (1988-1993), Boediono sebagai Wakil Presiden (2009-2014) yang fokus pada stabilisasi ekonomi pasca-krisis.")
print("- Boediono (UI) dan Pratikno (UGM): Interaksi melalui acara diplomatik, seperti pertemuan dengan PM Belanda di mana Pratikno hadir sebagai rektor UGM.")
print("- Ahmad Heryawan (Unpad) dan Boediono (UI): Bertemu di acara nasional seperti seminar transformasi pertanian di IPB University.")
print("- Ernest Prakasa (Unpad) dan Ridwan Kamil (ITB): Interaksi publik melalui kritik Ernest terhadap Ridwan Kamil di media sosial terkait kampanye politik di Jakarta.")
print("Relasi ini ditambahkan sebagai edges baru, yang mengurangi jumlah connected components (dari 4 menjadi 1 atau 2 tergantung koneksi), menunjukkan bahwa meskipun afiliasi universitas membentuk kelompok dasar, interaksi lintas institusi melalui pemerintahan dan politik nasional dapat membentuk jembatan antar kelompok, mencerminkan dinamika jejaring sosial nyata di Indonesia.")

pos = {}
# Posisi untuk UGM
pos["Pratikno"] = (1, 5)
pos["Adrianus Mooy"] = (0, 4)
pos["J Soedrajad Djiwandono"] = (2, 4)

# Posisi untuk ITB
pos["Ignasius Jonan"] = (4, 5)
pos["Ridwan Kamil"] = (3, 4)
pos["Hetifah Sjaifudian"] = (5, 4)

# Posisi untuk UI
pos["Boediono"] = (1, 2)
pos["Jusuf Wanandi"] = (0, 1)
pos["Mohammad Hatta"] = (2, 1)

# Posisi untuk Unpad
pos["Ahmad Heryawan"] = (4, 2)
pos["Hikmat Kurnia"] = (3, 1)
pos["Ernest Prakasa"] = (5, 1)

# Warna node berdasarkan kelompok
colors = ['lightblue'] * 3 + ['lightgreen'] * 3 + ['lightyellow'] * 3 + ['lightcoral'] * 3

# Gambar graf
plt.figure(figsize=(12, 8))

# Gambar edges intra-univ dengan solid, inter-univ dengan dashed
intra_edges = [(u, v) for u, v in G.edges() if (u in ugm_alumni and v in ugm_alumni) or (u in itb_alumni and v in itb_alumni) or (u in ui_alumni and v in ui_alumni) or (u in unpad_alumni and v in unpad_alumni)]
inter_edges = [edge for edge in G.edges() if edge not in intra_edges]

nx.draw_networkx_edges(G, pos, edgelist=intra_edges, edge_color='gray', style='solid')
nx.draw_networkx_edges(G, pos, edgelist=inter_edges, edge_color='red', style='dashed', width=2)

nx.draw_networkx_nodes(G, pos, node_color=colors, node_size=2000)
nx.draw_networkx_labels(G, pos, font_size=10, font_weight='bold')

# Tambahkan legenda untuk menunjukkan warna universitas
legend_elements = [
    Line2D([0], [0], marker='o', color='w', label='UGM', markerfacecolor='lightblue', markersize=15),
    Line2D([0], [0], marker='o', color='w', label='ITB', markerfacecolor='lightgreen', markersize=15),
    Line2D([0], [0], marker='o', color='w', label='UI', markerfacecolor='lightyellow', markersize=15),
    Line2D([0], [0], marker='o', color='w', label='Unpad', markerfacecolor='lightcoral', markersize=15),
]
plt.legend(handles=legend_elements, loc='upper right', title="Universitas")

plt.show()
```

  - Jelaskan makna Strongly Connected Components dalam konteks skenario ini.


```python
import networkx as nx
import matplotlib.pyplot as plt

# Fungsi untuk memvisualisasikan graf
def visualize_graph(G, title, pos=None):
    plt.figure(figsize=(8, 6))
    if pos is None:
        pos = nx.spring_layout(G, seed=42) # Gunakan spring_layout untuk tata letak

    # Menggambar simpul
    nx.draw_networkx_nodes(G, pos, node_color='skyblue', node_size=700)

    # Menggambar sisi (dengan panah untuk graf berarah)
    nx.draw_networkx_edges(G, pos, edge_color='gray', arrowstyle='->', arrowsize=20)

    # Menggambar label
    nx.draw_networkx_labels(G, pos, font_size=12, font_family="sans-serif")

    plt.title(title)
    plt.axis('off')
    plt.show()

# --- Task 4: Alur Komando dalam Dinamika Demonstrasi ---
print("--- Task 4: Alur Komando dalam Dinamika Demonstrasi ---")

# Modelkan skenario sebagai Directed Graph
G_komando = nx.DiGraph()

# Tentukan simpul-simpul
nodes = [
    'Juru Bicara Mahasiswa',
    'Koordinator Lapangan Mahasiswa',
    'Kepala Polisi',
    'Komandan Pasukan Polisi',
    'Komandan TNI',
    'Komandan Pasukan TNI',
    'Juru Bicara Pemerintah',
    'Menteri Terkait',
    'Provokator' # Menambahkan aktor tambahan
]
G_komando.add_nodes_from(nodes)

# Tambahkan sisi (alur komando/informasi satu arah)
G_komando.add_edges_from([
    ('Juru Bicara Pemerintah', 'Menteri Terkait'),
    ('Menteri Terkait', 'Kepala Polisi'),
    ('Menteri Terkait', 'Komandan TNI'),
    ('Kepala Polisi', 'Komandan Pasukan Polisi'),
    ('Komandan TNI', 'Komandan Pasukan TNI'),
    ('Juru Bicara Mahasiswa', 'Koordinator Lapangan Mahasiswa'),
    ('Provokator', 'Koordinator Lapangan Mahasiswa'), # Provokator mempengaruhi koordinator
    ('Koordinator Lapangan Mahasiswa', 'Juru Bicara Mahasiswa'), # Informasi dari lapangan ke juru bicara (tidak selalu perintah)
    ('Komandan Pasukan Polisi', 'Kepala Polisi'), # Laporan dari pasukan ke kepala polisi

    # Menambahkan sisi agar graf terhubung lemah
    ('Koordinator Lapangan Mahasiswa', 'Kepala Polisi'), # Mahasiswa ke Polisi
    ('Komandan Pasukan Polisi', 'Juru Bicara Pemerintah'), # Polisi ke Pemerintah (laporan)
    ('Juru Bicara Pemerintah', 'Juru Bicara Mahasiswa') # Pemerintah ke Mahasiswa (negosiasi/pesan)
])

visualize_graph(G_komando, "Alur Komando dan Informasi dalam Demonstrasi (Updated)")

# Buktikan bahwa graf tersebut tidak terhubung kuat
print("\nAnalisis Keterhubungan:")
is_strongly_connected = nx.is_strongly_connected(G_komando)
is_weakly_connected = nx.is_weakly_connected(G_komando)

print(f"Apakah graf terhubung kuat? {is_strongly_connected}")
print(f"Apakah graf terhubung lemah? {is_weakly_connected}")

# Identifikasi jalur komando yang terputus
print("\nJalur Komando yang Terputus (Bukti tidak terhubung kuat):")
if not is_strongly_connected:
    print("Graf tidak terhubung kuat karena tidak setiap simpul dapat dijangkau dari setiap simpul lainnya.")
    print("Contoh jalur yang terputus:")
    print("- Tidak ada jalur yang memungkinkan perintah langsung dari Kepala Polisi ke Juru Bicara Mahasiswa.")
    print("- Tidak ada jalur yang memungkinkan alur komando formal dari Juru Bicara Mahasiswa kembali ke Juru Bicara Pemerintah melalui rantai komando yang ada.")
    print("Ini menunjukkan bahwa alur komando dan informasi bersifat hierarkis dan tidak selalu memungkinkan komunikasi dua arah antara semua pihak secara langsung atau melalui rantai komando/informasi yang ada.")


# Temukan dan tampilkan Strongly Connected Components
strong_components = list(nx.strongly_connected_components(G_komando))
print(f"\nStrongly Connected Components: {strong_components}")

# Jelaskan makna Strongly Connected Components
print("\nMakna Strongly Connected Components dalam konteks skenario ini:")
print("Strongly Connected Components (SCC) merepresentasikan subkelompok simpul di mana setiap simpul dalam subkelompok tersebut dapat dijangkau dari setiap simpul lainnya dalam subkelompok yang sama melalui jalur berarah.")
print("Dalam skenario alur komando ini, SCC mengidentifikasi kelompok aktor yang memiliki komunikasi dua arah atau siklus informasi/perintah di antara mereka.")
print("Contoh: Jika sebuah SCC terdiri dari 'Kepala Polisi' dan 'Komandan Pasukan Polisi', ini berarti bahwa ada alur informasi/perintah dari Kepala Polisi ke Komandan Pasukan, dan juga ada laporan atau informasi yang mengalir kembali dari Komandan Pasukan ke Kepala Polisi, membentuk lingkaran komunikasi internal.")
print("SCC yang hanya berisi satu simpul (seperti {'Juru Bicara Pemerintah'}) menunjukkan bahwa simpul tersebut mungkin memiliki alur keluar (memberi perintah/informasi) tetapi tidak ada alur kembali yang terhubung kuat dalam konteks SCC tersebut.")
```
